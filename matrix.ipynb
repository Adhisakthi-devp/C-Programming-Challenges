```python
import pandas as pd
import matplotlib.pyplot as plt
import time
from IPython.display import clear_output

# Load the smooth CSV
df = pd.read_csv(r"C:\Users\Adhi Sakthi S\Documents\Bird sensing\modified_sensor_data.csv")

# Scale factor (to make graph values more visible)
scale_factor = 1.2  # Increase/decrease as you like

# Continuous plot simulation
for i in range(1, len(df)):
    current_data = df.iloc[:i]

    clear_output(wait=True)  # Clear old plot
    fig, axs = plt.subplots(3, 2, figsize=(14, 12))  # 3 rows × 2 columns
    axs = axs.ravel()  # flatten array for easy indexing

    # Gas Sensor
    axs[0].plot(current_data["Time(ms)"], current_data["Gas"] * scale_factor, 
                color="purple", marker="o", linewidth=2)
    axs[0].set_title("Gas Sensor")
    axs[0].grid(True)

    # Temperatures
    axs[1].plot(current_data["Time(ms)"], current_data["DHT_Temp"] * scale_factor, label="DHT Temp", color="red")
    axs[1].plot(current_data["Time(ms)"], current_data["BMP_Temp"] * scale_factor, label="BMP Temp", color="orange")
    axs[1].plot(current_data["Time(ms)"], current_data["MPU_Temp"] * scale_factor, label="MPU Temp", color="brown")
    axs[1].legend()
    axs[1].set_title("Temperatures")
    axs[1].grid(True)

    # Humidity
    axs[2].plot(current_data["Time(ms)"], current_data["DHT_Hum"] * scale_factor, 
                color="blue", linestyle="--", linewidth=2)
    axs[2].set_title("Humidity")
    axs[2].grid(True)

    # Pressure
    axs[3].plot(current_data["Time(ms)"], current_data["BMP_Press"] * scale_factor, 
                color="green", linewidth=2)
    axs[3].set_title("Pressure")
    axs[3].grid(True)

    # ADXL Accelerometer
    axs[4].plot(current_data["Time(ms)"], current_data["ADXL_X"] * scale_factor, label="X", color="cyan")
    axs[4].plot(current_data["Time(ms)"], current_data["ADXL_Y"] * scale_factor, label="Y", color="magenta")
    axs[4].plot(current_data["Time(ms)"], current_data["ADXL_Z"] * scale_factor, label="Z", color="black")
    axs[4].legend()
    axs[4].set_title("ADXL Accelerometer")
    axs[4].grid(True)

    # MPU Accelerometer Z
    axs[5].plot(current_data["Time(ms)"], current_data["MPU_Az"] * scale_factor, 
                color="darkblue", linewidth=2)
    axs[5].set_title("MPU Accelerometer Z")
    axs[5].grid(True)

    plt.suptitle("Dynamic Sensor Data Visualization", fontsize=16, fontweight="bold")
    plt.tight_layout(rect=[0, 0, 1, 0.97])
    plt.show()

    time.sleep(0.2)  # Delay for animation
```
